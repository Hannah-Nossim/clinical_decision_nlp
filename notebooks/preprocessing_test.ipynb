```python
import pandas as pd
import re
```

```python
test_df = pd.read_csv("../data/test_raw.csv", encoding='utf-8', on_bad_lines='warn')
test_df.head()
```

```python
def clean_text_advanced(text):
    if pd.isna(text):
        return ""
    text = str(text)
    text = re.sub(r'_x000D_|_x0001_|\\n|\\t', ' ', text)
    text = re.sub(r'\s+', ' ', text)
    return text.strip()

for col in ['Prompt', 'Clinical Panel', 'County']:
    if col in test_df.columns:
        test_df[f"{col}_clean"] = test_df[col].apply(clean_text_advanced)
    else:
        print(f"Warning: Column '{col}' not found in test dataset.")

test_df[['Prompt_clean', 'Clinical Panel_clean']].head()

```

```python
test_structured = pd.DataFrame()
test_structured['input_text'] = "CLINICAL CASE: " + test_df['Prompt_clean']
test_structured['target_text'] = ""  
test_structured['master_index'] = test_df['Master_Index']
test_structured['clinical_panel'] = test_df['Clinical Panel_clean']

test_structured.head()

```

```python
output_path = "../data/test_structured_for_eval.csv"
test_structured.to_csv(output_path, index=False, encoding='utf-8')

print(f"Preprocessed test dataset saved to: {output_path}")
print(f"Shape: {test_structured.shape}")

```
